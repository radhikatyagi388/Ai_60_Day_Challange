# Day 6 - What is Machine Learning?

## Experiment
I used Google's Teachable Machine to train an image classifier
with two classes: Pen and Mobile.

## Experiment 1: Training Data
I trained the model with increasing amounts of training data
and observed how the predictions changed.

## Experiment 2: Noisy Data
I added inconsistent images and tested how this affected
the model's predictions.

## Observations

1. More diverse training data generally improved prediction consistency.
2. Noisy and inconsistent data reduced model reliability.
3. Training data should represent the real-world situations
   in which the model will be used.

## Key Learning

Machine learning models learn patterns from examples.
The quality, quantity, diversity, and balance of training data
strongly influence how well the model performs on new inputs.
