```python
import nltk
nltk.download('punkt')
nltk.download('punkt_tab')
nltk.download('stopwords')

from nltk.tokenize import word_tokenize
from nltk.corpus import stopwords
import string
```

```python
text=input("enter your text: ")
```

```python
print(text)
```

```python
tokens=word_tokenize(text)
print("tokens:")
print(tokens)
```

```python
lowercase_tokens=[word.lower() for word in tokens]
print(lowercase_tokens)
```

```python
clean_tokens=[]

for word in lowercase_tokens:
  if not any(char in string.punctuation for char in word):
         clean_tokens.append(word)

print("after removing punctuation: ")
print(clean_tokens)
```

```python
stop_words=set(stopwords.words('english'))
filtered_tokens=[]

for word in clean_tokens:
  if word not in stop_words:
    filtered_tokens.append(word)

  print("after removing stopwords: ")
  print(filtered_tokens)
```

```python
word_counts={}
for word in filtered_tokens:
  if word in word_counts:
    word_counts[word]+=1
  else:
    word_counts[word]=1
print("Bag of words: ")
for word, count in word_counts.items():
      print(f"{word}:{count}")
```

```python
vocabulary=sorted(word_counts.keys())
print("vocabulary: ")
print(vocabulary)
```

```python
bow_vector=[word_counts[word] for word in vocabulary]
print("Bag of word vector: ")
print(bow_vector)
```

```python
print("Step 1 - original text")
print(text)
print("\nStep 2 - tokenization")
print(tokens)
print("\nStep3 - Lowercase")
print(lowercase_tokens)
print("\nStep4 - remove punctuatiom")
print(clean_tokens)
print("\nStep5 - remove stop words")
print(filtered_tokens)
print("\nStep6- bag of words")
print(word_counts)
print("\nStep7 - numerical vector")
print(bow_vector)
```

```python
import pandas as pd
dataset=pd.DataFrame({
    "original_text": text,
    "processed_text":[
        " ".join([
            word for word in word_tokenize(text.lower())
            if word not in string.punctuation
            and word not in stop_words
        ])
        for text in text
    ]
})
dataset
```
