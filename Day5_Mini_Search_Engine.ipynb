```python
from sentence_transformers import SentenceTransformer
from sklearn.metrics.pairwise import cosine_similarity
import numpy as np
import pandas as pd
```

```python
documents = [
    "Artificial intelligence enables computers to perform tasks that normally require human intelligence.",
    "Machine learning allows computers to learn patterns from data without being explicitly programmed.",
    "Deep learning uses neural networks with many layers to learn complex patterns.",
    "Natural language processing helps computers understand and generate human language.",
    "Computer vision allows machines to understand images and videos.",
    "Reinforcement learning trains an agent through rewards and penalties.",
    "Generative AI can create text, images, music, and other forms of content.",
    "Large language models can generate human-like responses from natural language prompts.",
    "Chatbots use artificial intelligence to communicate with users through conversations.",
    "Recommendation systems suggest products or content based on user behavior.",
    "Semantic search finds information based on meaning rather than exact keywords.",
    "Embeddings represent text as numerical vectors that capture semantic meaning.",
    "Vector databases store embeddings and allow fast similarity searches.",
    "Retrieval augmented generation combines information retrieval with language generation.",
    "AI agents can plan tasks and use tools to achieve specific goals.",
    "Training data is used to teach machine learning models how to recognize patterns.",
    "A neural network consists of interconnected nodes inspired by biological neurons.",
    "AI models can be evaluated using metrics such as accuracy, precision, and recall.",
    "Prompt engineering involves designing instructions that guide AI models toward useful outputs.",
    "AI applications are increasingly used in healthcare, education, finance, and entertainment."
]
```

```python
model = SentenceTransformer("sentence-transformers/all-MiniLM-L6-v2")
```

```python
embeddings = model.encode(documents)

print("Number of sentences:", len(documents))
print("Embedding shape:", embeddings.shape)
```

```python
def semantic_search(query, top_k=3):

    # Convert query into embedding
    query_embedding = model.encode([query])

    # Calculate similarity with all documents
    scores = cosine_similarity(query_embedding, embeddings)[0]

    # Get indexes of top results
    top_indices = np.argsort(scores)[::-1][:top_k]

    results = []

    for index in top_indices:
        results.append((documents[index], scores[index]))

    return results
```

```python
query = "How do computers understand human language?"

results = semantic_search(query)

for i, (text, score) in enumerate(results, 1):
    print(f"{i}. {text}")
    print(f"   Similarity: {score:.4f}\n")
```

```python
def keyword_search(query, top_k=3):

    query_words = set(query.lower().split())

    results = []

    for doc in documents:
        doc_words = set(doc.lower().split())

        # Count matching words
        score = len(query_words.intersection(doc_words))

        results.append((doc, score))

    results.sort(key=lambda x: x[1], reverse=True)

    return results[:top_k]
```

```python
query = "How can machines understand what people say?"

print("QUERY:")
print(query)

print("\n--- KEYWORD SEARCH ---")

keyword_results = keyword_search(query)

for i, (text, score) in enumerate(keyword_results, 1):
    print(f"{i}. {text}")
    print(f"   Keyword score: {score}")

print("\n--- SEMANTIC SEARCH ---")

semantic_results = semantic_search(query)

for i, (text, score) in enumerate(semantic_results, 1):
    print(f"{i}. {text}")
    print(f"   Semantic score: {score:.4f}")
```

```python
query = "Finding information based on meaning"
```

```python
def comparison_table(query):

    keyword_results = keyword_search(query)
    semantic_results = semantic_search(query)

    data = []

    for rank, (text, score) in enumerate(keyword_results, 1):
        data.append({
            "Search Type": "Keyword",
            "Rank": rank,
            "Result": text,
            "Score": score
        })

    for rank, (text, score) in enumerate(semantic_results, 1):
        data.append({
            "Search Type": "Semantic",
            "Rank": rank,
            "Result": text,
            "Score": round(float(score), 4)
        })

    return pd.DataFrame(data)
```

```python
comparison_table("Finding information based on meaning")
```

```python
queries = [
    "How do machines understand human language?",
    "How does AI recommend things to users?",
    "How can AI create new content?",
    "How are images understood by computers?",
    "How can AI retrieve relevant information?"
]

for query in queries:

    print("=" * 70)
    print("QUERY:", query)
    print("=" * 70)

    results = semantic_search(query)

    for i, (text, score) in enumerate(results, 1):
        print(f"{i}. {text}")
        print(f"   Score: {score:.4f}")

    print()
```

```python
## Semantic Search vs Keyword Search

#Keyword search looks for exact words appearing in the query and documents.
#It can fail when the user uses different words with the same meaning.

#Semantic search converts text into vector embeddings and compares the meaning
#of the query with the meaning of each document using cosine similarity.

#For example, the query "How can machines understand what people say?"
#may not contain the exact words "natural language processing", but semantic
#search can still identify NLP-related sentences as relevant.

#This is the basic retrieval mechanism used in modern AI systems and is an
#important component of Retrieval-Augmented Generation (RAG).
```

## Key Learnings

1. Text can be converted into numerical embeddings using an embedding model.
2. Embeddings capture semantic relationships between sentences.
3. Cosine similarity can be used to measure similarity between embeddings.
4. Semantic search can find relevant information even when exact keywords are different.
5. Keyword search depends heavily on matching words, while semantic search focuses on meaning.
6. This retrieval process is a fundamental building block of RAG systems.
