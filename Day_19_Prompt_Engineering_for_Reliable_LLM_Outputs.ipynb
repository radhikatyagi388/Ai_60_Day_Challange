```python
# ================================================================
# DAY 19 - PROMPT ENGINEERING FOR RELIABLE LLM OUTPUTS
# NO OPENAI API KEY REQUIRED
# Uses Hugging Face Transformers + Qwen local model
# ================================================================

# -------------------------------
# 1. INSTALL LIBRARIES
# -------------------------------

!pip install -q transformers accelerate sentencepiece pandas matplotlib torch


# -------------------------------
# 2. IMPORTS
# -------------------------------

import torch
import json
import pandas as pd
import matplotlib.pyplot as plt

from transformers import AutoTokenizer, AutoModelForCausalLM


# -------------------------------
# 3. LOAD LOCAL MODEL
# -------------------------------

MODEL_NAME = "Qwen/Qwen2.5-0.5B-Instruct"

print("Loading model...")

tokenizer = AutoTokenizer.from_pretrained(MODEL_NAME)

model = AutoModelForCausalLM.from_pretrained(
    MODEL_NAME,
    torch_dtype="auto",
    device_map="auto"
)

print("Model loaded successfully!")


# -------------------------------
# 4. FUNCTION TO GENERATE ANSWER
# -------------------------------

def ask_model(prompt, max_tokens=150):

    messages = [
        {
            "role": "user",
            "content": prompt
        }
    ]

    text = tokenizer.apply_chat_template(
        messages,
        tokenize=False,
        add_generation_prompt=True
    )

    inputs = tokenizer(
        text,
        return_tensors="pt"
    ).to(model.device)

    with torch.no_grad():

        outputs = model.generate(
            **inputs,
            max_new_tokens=max_tokens,
            do_sample=False,
            temperature=None
        )

    generated = outputs[0][inputs["input_ids"].shape[1]:]

    answer = tokenizer.decode(
        generated,
        skip_special_tokens=True
    )

    return answer.strip()


# -------------------------------
# 5. 10 TEST CASES
# -------------------------------

test_cases = [

    {
        "id": 1,
        "context":
        "Python was created by Guido van Rossum and first released in 1991.",
        "question":
        "Who created Python?",
        "answer":
        "Guido van Rossum"
    },

    {
        "id": 2,
        "context":
        "The Amazon rainforest is primarily located in Brazil. It is the largest tropical rainforest in the world.",
        "question":
        "Where is most of the Amazon rainforest located?",
        "answer":
        "Brazil"
    },

    {
        "id": 3,
        "context":
        "The Earth revolves around the Sun. One complete revolution takes approximately 365.25 days.",
        "question":
        "How long does Earth take to orbit the Sun?",
        "answer":
        "365.25 days"
    },

    {
        "id": 4,
        "context":
        "HTTP stands for HyperText Transfer Protocol. It is used for communication between web browsers and servers.",
        "question":
        "What does HTTP stand for?",
        "answer":
        "HyperText Transfer Protocol"
    },

    {
        "id": 5,
        "context":
        "The human heart has four chambers: two atria and two ventricles.",
        "question":
        "How many chambers does the human heart have?",
        "answer":
        "four"
    },

    {
        "id": 6,
        "context":
        "Mount Everest is the highest mountain above sea level. Its elevation is approximately 8,849 meters.",
        "question":
        "What is the highest mountain above sea level?",
        "answer":
        "Mount Everest"
    },

    {
        "id": 7,
        "context":
        "JavaScript is a programming language commonly used to make web pages interactive.",
        "question":
        "What is JavaScript commonly used for?",
        "answer":
        "making web pages interactive"
    },

    {
        "id": 8,
        "context":
        "Water freezes at 0 degrees Celsius under standard atmospheric pressure.",
        "question":
        "At what temperature does water freeze?",
        "answer":
        "0 degrees Celsius"
    },

    {
        "id": 9,
        "context":
        "The Pacific Ocean is the largest ocean on Earth.",
        "question":
        "What is the largest ocean on Earth?",
        "answer":
        "Pacific Ocean"
    },

    {
        "id": 10,
        "context":
        "HTML stands for HyperText Markup Language. It is used to structure content on web pages.",
        "question":
        "What does HTML stand for?",
        "answer":
        "HyperText Markup Language"
    }
]


# -------------------------------
# 6. PROMPT VERSIONS
# -------------------------------

prompts = {

    "baseline": """
Answer the question using the provided context.

Context:
{context}

Question:
{question}

Answer:
""",

    "role_assignment": """
You are a precise document question-answering assistant.

Answer the question using the provided context.

Context:
{context}

Question:
{question}

Answer:
""",

    "output_format": """
Answer the question using only the provided context.

Return exactly this format:

Answer: <short answer>

Context:
{context}

Question:
{question}
""",

    "reasoning": """
You are a precise document question-answering assistant.

Carefully reason about the context before answering.
Use only information supported by the context.
Do not show your reasoning.
Return only the final answer.

Context:
{context}

Question:
{question}

Answer:
""",

    "few_shot": """
You are a precise document question-answering assistant.

Follow these examples.

Example 1:
Context: Paris is the capital of France.
Question: What is the capital of France?
Answer: Paris

Example 2:
Context: The Pacific Ocean is the largest ocean.
Question: What is the largest ocean?
Answer: Pacific Ocean

Now answer the following question.

Context:
{context}

Question:
{question}

Answer:
""",

    "negative_constraints": """
Answer the question using only the provided context.

Do not:
- use outside knowledge
- invent information
- guess
- add unsupported facts
- provide unnecessary explanations

If the answer is not present in the context, say:
Not stated in the context.

Return only the answer.

Context:
{context}

Question:
{question}

Answer:
"""
}


# -------------------------------
# 7. SIMPLE AUTOMATIC SCORING
# -------------------------------
# Instead of using another LLM as evaluator,
# we compare the generated answer with the expected answer.
#
# This avoids needing a second API/model.
# -------------------------------

def normalize(text):

    return (
        text.lower()
        .replace(".", "")
        .replace(",", "")
        .replace(":", "")
        .replace('"', "")
        .strip()
    )


def calculate_accuracy(expected, actual):

    expected = normalize(expected)
    actual = normalize(actual)

    if expected in actual:
        return 5

    expected_words = set(expected.split())
    actual_words = set(actual.split())

    if len(expected_words) == 0:
        return 1

    overlap = len(
        expected_words.intersection(actual_words)
    ) / len(expected_words)

    if overlap >= 0.75:
        return 4

    elif overlap >= 0.5:
        return 3

    elif overlap > 0:
        return 2

    else:
        return 1


def calculate_format(version, answer):

    answer = answer.strip()

    if len(answer) == 0:
        return 1

    # Strict format requirement for output_format
    if version == "output_format":

        if answer.lower().startswith("answer:"):
            return 5
        else:
            return 2

    # Other prompts should return concise answers
    if len(answer.split()) <= 15:
        return 5

    elif len(answer.split()) <= 30:
        return 4

    elif len(answer.split()) <= 50:
        return 3

    else:
        return 2


# -------------------------------
# 8. RUN ALL PROMPTS
# -------------------------------

all_results = []

for version_name, prompt in prompts.items():

    print("\n" + "=" * 60)
    print("Testing:", version_name)
    print("=" * 60)

    for item in test_cases:

        formatted_prompt = prompt.format(
            context=item["context"],
            question=item["question"]
        )

        answer = ask_model(formatted_prompt)

        accuracy = calculate_accuracy(
            item["answer"],
            answer
        )

        format_score = calculate_format(
            version_name,
            answer
        )

        all_results.append({

            "version": version_name,

            "id": item["id"],

            "question": item["question"],

            "expected_answer": item["answer"],

            "model_answer": answer,

            "accuracy": accuracy,

            "format_consistency": format_score
        })

        print(
            f"Test {item['id']} | "
            f"Accuracy: {accuracy}/5 | "
            f"Format: {format_score}/5"
        )

        print("Answer:", answer)


# -------------------------------
# 9. RESULTS DATAFRAME
# -------------------------------

results_df = pd.DataFrame(all_results)

print("\n\nDETAILED RESULTS")
display(results_df)


# -------------------------------
# 10. CALCULATE AVERAGES
# -------------------------------

summary = (
    results_df
    .groupby("version")
    [["accuracy", "format_consistency"]]
    .mean()
)

summary["overall_score"] = (
    summary["accuracy"] +
    summary["format_consistency"]
) / 2


# -------------------------------
# 11. BASELINE IMPROVEMENT
# -------------------------------

baseline_score = summary.loc[
    "baseline",
    "overall_score"
]

summary["improvement_vs_baseline"] = (
    summary["overall_score"] -
    baseline_score
)

summary = summary.sort_values(
    "overall_score",
    ascending=False
)

print("\n\nPROMPT COMPARISON")
display(summary.round(2))


# -------------------------------
# 12. PROMPT VERSION DICTIONARY
# -------------------------------

prompt_versions = {

    "baseline": {
        "average_score": None,
        "description":
        "Simple question-answering prompt."
    },

    "role_assignment": {
        "average_score": None,
        "description":
        "Added a precise document-QA assistant role."
    },

    "output_format": {
        "average_score": None,
        "description":
        "Added an explicit answer format."
    },

    "reasoning": {
        "average_score": None,
        "description":
        "Added careful reasoning before producing the final answer."
    },

    "few_shot": {
        "average_score": None,
        "description":
        "Added examples showing the desired question-answer behavior."
    },

    "negative_constraints": {
        "average_score": None,
        "description":
        "Added restrictions against guessing and unsupported information."
    }
}


for version in prompt_versions:

    prompt_versions[version][
        "average_score"
    ] = round(
        summary.loc[
            version,
            "overall_score"
        ],
        2
    )


print("\n\nPROMPT VERSION DICTIONARY")

for version, data in prompt_versions.items():

    print(
        f"\n{version}: "
        f"{data['average_score']}/5"
    )

    print(
        "Change:",
        data["description"]
    )


# -------------------------------
# 13. FIND BEST TECHNIQUE
# -------------------------------

techniques = [
    "role_assignment",
    "output_format",
    "reasoning",
    "few_shot",
    "negative_constraints"
]

best_technique = max(
    techniques,
    key=lambda x:
    prompt_versions[x]["average_score"]
)

best_score = prompt_versions[
    best_technique
]["average_score"]

largest_improvement = summary.loc[
    techniques,
    "improvement_vs_baseline"
].idxmax()

improvement_value = summary.loc[
    largest_improvement,
    "improvement_vs_baseline"
]


# -------------------------------
# 14. HYPOTHESIS
# -------------------------------

hypotheses = {

    "role_assignment":
    "Role assignment may improve performance by giving the model a clearer task identity.",

    "output_format":
    "Output format specification may improve consistency by reducing variation in how answers are presented.",

    "reasoning":
    "Reasoning instructions may improve accuracy by encouraging the model to inspect the context carefully.",

    "few_shot":
    "Few-shot examples may work best because they demonstrate the exact behavior expected from the model.",

    "negative_constraints":
    "Negative constraints may work best because they reduce guessing, unsupported claims, and irrelevant information."
}


print("\n\n" + "=" * 60)
print("BEST TECHNIQUE ANALYSIS")
print("=" * 60)

print("Best technique:", best_technique)
print("Best average score:", best_score)
print(
    "Largest improvement:",
    round(improvement_value, 2)
)

print("\nHypothesis:")
print(hypotheses[best_technique])


# ================================================================
# 15. GROUNDING SYSTEM PROMPT
# ================================================================

grounding_prompt = """
You are a grounded RAG question-answering assistant.

Your ONLY source of information is the CONTEXT.

Rules:

1. Answer only using information explicitly supported by the context.
2. Do not use pretrained knowledge.
3. Do not guess or speculate.
4. Do not invent facts.
5. If the answer is not supported by the context,
   refuse to answer.
6. When refusing, respond exactly:

I cannot answer this question from the provided context.

7. Never answer from general knowledge when the context is insufficient.
"""


# -------------------------------
# 16. FIVE OUT-OF-CONTEXT QUESTIONS
# -------------------------------

grounding_tests = [

    {
        "id": 1,
        "context":
        "Python was created by Guido van Rossum and first released in 1991.",
        "question":
        "What is the current market value of Python?"
    },

    {
        "id": 2,
        "context":
        "The Pacific Ocean is the largest ocean on Earth.",
        "question":
        "Who discovered the Pacific Ocean?"
    },

    {
        "id": 3,
        "context":
        "HTTP stands for HyperText Transfer Protocol.",
        "question":
        "Who invented HTTP?"
    },

    {
        "id": 4,
        "context":
        "The human heart has four chambers.",
        "question":
        "What is the average resting heart rate of a professional athlete?"
    },

    {
        "id": 5,
        "context":
        "Mount Everest is approximately 8,849 meters high.",
        "question":
        "How many people have successfully climbed Mount Everest?"
    }
]


# -------------------------------
# 17. RUN GROUNDING TESTS
# -------------------------------

grounding_results = []

for test in grounding_tests:

    prompt = f"""
{grounding_prompt}

CONTEXT:
{test['context']}

QUESTION:
{test['question']}

ANSWER:
"""

    answer = ask_model(
        prompt,
        max_tokens=100
    )

    refused = (
        "cannot answer" in answer.lower()
        or
        "not stated" in answer.lower()
        or
        "not provided" in answer.lower()
        or
        "not enough information" in answer.lower()
    )

    grounding_results.append({

        "test": test["id"],

        "question": test["question"],

        "model_response": answer,

        "refused_correctly": refused
    })


# -------------------------------
# 18. GROUNDING RESULTS
# -------------------------------

grounding_df = pd.DataFrame(
    grounding_results
)

print("\n\n" + "=" * 60)
print("GROUNDING TEST RESULTS")
print("=" * 60)

display(grounding_df)


grounding_accuracy = (
    grounding_df["refused_correctly"]
    .mean() * 100
)

print(
    f"\nGrounding refusal accuracy: "
    f"{grounding_accuracy:.1f}%"
)


# -------------------------------
# 19. GROUNDING TEST DETAILS
# -------------------------------

for _, row in grounding_df.iterrows():

    if row["refused_correctly"]:
        status = "CORRECT REFUSAL"
    else:
        status = "HALLUCINATION / ANSWERED FROM KNOWLEDGE"

    print("\nTest:", row["test"])
    print("Question:", row["question"])
    print("Response:", row["model_response"])
    print("Result:", status)


# -------------------------------
# 20. CHART 1
# -------------------------------

plt.figure(figsize=(10, 5))

summary["overall_score"].plot(
    kind="bar"
)

plt.title(
    "Prompt Engineering Technique Comparison"
)

plt.xlabel("Prompt Version")

plt.ylabel("Average Score")

plt.ylim(0, 5)

plt.xticks(rotation=45)

plt.tight_layout()

plt.show()


# -------------------------------
# 21. CHART 2
# -------------------------------

plt.figure(figsize=(10, 5))

summary[
    ["accuracy", "format_consistency"]
].plot(
    kind="bar"
)

plt.title(
    "Accuracy vs Format Consistency"
)

plt.xlabel("Prompt Version")

plt.ylabel("Score")

plt.ylim(0, 5)

plt.xticks(rotation=45)

plt.tight_layout()

plt.show()


# -------------------------------
# 22. FINAL COMPARISON TABLE
# -------------------------------

final_comparison = summary.reset_index()

final_comparison.columns = [

    "Prompt Version",

    "Accuracy",

    "Format Consistency",

    "Overall Score",

    "Improvement vs Baseline"
]

print("\n\nFINAL COMPARISON TABLE")

display(
    final_comparison.round(2)
)


# -------------------------------
# 23. SAVE CSV FILES
# -------------------------------

results_df.to_csv(
    "prompt_engineering_results.csv",
    index=False
)

final_comparison.to_csv(
    "prompt_comparison.csv",
    index=False
)

grounding_df.to_csv(
    "grounding_results.csv",
    index=False
)


# -------------------------------
# 24. FINAL REPORT
# -------------------------------

print("\n")
print("=" * 70)
print("DAY 19 - FINAL REPORT")
print("=" * 70)

print("\nTask:")
print("Document Question Answering")

print("\nTest Inputs:")
print(len(test_cases))

print("\nPrompt Versions:")
print(len(prompts))

print("\nBest Technique:")
print(best_technique)

print("\nBest Average Score:")
print(best_score)

print("\nLargest Improvement:")
print(round(improvement_value, 2))

print("\nGrounding Refusal Accuracy:")
print(f"{grounding_accuracy:.1f}%")

print("\nHypothesis:")
print(hypotheses[best_technique])

print("\nGrounding System Prompt:")
print(grounding_prompt)

print("\nSaved Files:")
print("1. prompt_engineering_results.csv")
print("2. prompt_comparison.csv")
print("3. grounding_results.csv")

print("\nDAY 19 EXPERIMENT COMPLETED!")
```
