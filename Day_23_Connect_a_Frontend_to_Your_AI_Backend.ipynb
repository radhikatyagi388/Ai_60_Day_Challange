````python
# 🚀 Day 23 — Full-Stack AI Integration

## 📌 Project Overview

Built a complete **Full-Stack AI Assistant** that connects a responsive Next.js/React frontend with a FastAPI backend.

The project demonstrates how an AI backend can be transformed from a developer-only API into a usable web application for non-technical users.

---

## 🎯 Objective

The goal of this project was to build a functional AI chat application with:

- Next.js frontend
- React chat interface
- FastAPI backend
- Streaming AI responses
- Markdown rendering
- Source citations
- Loading states
- Error handling
- Persistent multi-turn conversation
- Mobile-responsive UI

---

## 🏗️ System Architecture

```text
                    USER
                      │
                      ▼
             ┌─────────────────┐
             │     Chrome      │
             │  Responsive UI  │
             └────────┬────────┘
                      │
                      │ POST /api/ask
                      ▼
             ┌─────────────────┐
             │     Next.js     │
             │   React Frontend│
             │                 │
             │  Chat Interface │
             │  ReactMarkdown  │
             │  Loading State  │
             │  Error Handling │
             │  Source Panel   │
             └────────┬────────┘
                      │
                      │ Server-side Proxy
                      ▼
             ┌─────────────────┐
             │     FastAPI     │
             │                 │
             │    POST /ask    │
             └────────┬────────┘
                      │
                      ▼
             ┌─────────────────┐
             │ Answer Generator│
             └────────┬────────┘
                      │
                      ▼
             ┌─────────────────┐
             │ StreamingResponse│
             │    NDJSON        │
             └────────┬────────┘
                      │
                      ▼
             ┌─────────────────┐
             │ ReadableStream  │
             │   in Browser    │
             └────────┬────────┘
                      │
                      ▼
             ┌─────────────────┐
             │ ReactMarkdown   │
             │ + Sources       │
             └─────────────────┘
````

````python
%%writefile /content/fullstack-ai/backend/main.py

from fastapi import FastAPI, HTTPException
from fastapi.middleware.cors import CORSMiddleware
from pydantic import BaseModel
from fastapi.responses import StreamingResponse
import asyncio
import json


# ============================================================
# CREATE FASTAPI APP
# ============================================================

app = FastAPI(
    title="Full Stack AI Assistant",
    version="1.0.0"
)


# ============================================================
# CORS CONFIGURATION
# ============================================================

app.add_middleware(
    CORSMiddleware,
    allow_origins=["*"],
    allow_credentials=False,
    allow_methods=["*"],
    allow_headers=["*"],
)


# ============================================================
# REQUEST MODEL
# ============================================================

class AskRequest(BaseModel):
    question: str
    history: list = []


# ============================================================
# AI RESPONSE GENERATOR
# ============================================================

def generate_answer(question: str):

    q = question.lower()


    # --------------------------------------------------------
    # FASTAPI QUESTION
    # --------------------------------------------------------

    if "fastapi" in q:

        answer = "\n".join([
            "# FastAPI",
            "",
            "**FastAPI** is a modern Python framework for building APIs.",
            "",
            "### Why use FastAPI?",
            "",
            "- High performance",
            "- Automatic API documentation",
            "- Type validation using Pydantic",
            "- Easy asynchronous programming",
            "- Excellent support for AI applications",
            "",
            "### Example",
            "",
            "```python",
            "from fastapi import FastAPI",
            "",
            "app = FastAPI()",
            "",
            '@app.get("/")',
            "def home():",
            '    return {"message": "Hello AI"}',
            "```",
            "",
            "FastAPI can be used as the backend of an AI application.",
            "It receives requests from the frontend and sends responses back to the user."
        ])


        sources = [
            {
                "title": "FastAPI Documentation",
                "url": "https://fastapi.tiangolo.com/"
            },
            {
                "title": "Pydantic Documentation",
                "url": "https://docs.pydantic.dev/"
            }
        ]


    # --------------------------------------------------------
    # REACT / NEXT.JS QUESTION
    # --------------------------------------------------------

    elif "react" in q or "next.js" in q or "nextjs" in q:

        answer = "\n".join([
            "# Next.js and React",
            "",
            "**Next.js** is a React framework for building modern web applications.",
            "",
            "### Important features",
            "",
            "- Server-side rendering",
            "- Client-side rendering",
            "- File-based routing",
            "- API integration",
            "- Responsive applications",
            "",
            "### How this project works",
            "",
            "The user enters a question in the React interface.",
            "",
            "Next.js sends the question to the FastAPI backend using the **Fetch API**.",
            "",
            "```javascript",
            'fetch("/ask", {',
            '    method: "POST"',
            "})",
            "```",
            "",
            "FastAPI then streams the response back to the frontend."
        ])


        sources = [
            {
                "title": "Next.js Documentation",
                "url": "https://nextjs.org/docs"
            },
            {
                "title": "React Documentation",
                "url": "https://react.dev/"
            }
        ]


    # --------------------------------------------------------
    # AI QUESTION
    # --------------------------------------------------------

    elif "artificial intelligence" in q or q == "ai" or " ai " in q:

        answer = "\n".join([
            "# Artificial Intelligence",
            "",
            "**Artificial Intelligence (AI)** allows computers to perform tasks that normally require human intelligence.",
            "",
            "### Common AI applications",
            "",
            "- Natural language processing",
            "- Computer vision",
            "- Speech recognition",
            "- Recommendation systems",
            "- Generative AI",
            "",
            "### Full-Stack AI",
            "",
            "A full-stack AI application can contain:",
            "",
            "1. Frontend",
            "2. Backend API",
            "3. AI model",
            "4. Streaming system",
            "5. Source citations",
            "",
            "In this project, **Next.js** provides the frontend and **FastAPI** provides the backend."
        ])


        sources = [
            {
                "title": "Google AI",
                "url": "https://ai.google/"
            },
            {
                "title": "IBM Artificial Intelligence",
                "url": "https://www.ibm.com/think/topics/artificial-intelligence"
            }
        ]


    # --------------------------------------------------------
    # DEFAULT RESPONSE
    # --------------------------------------------------------

    else:

        answer = "\n".join([
            "# AI Assistant Response",
            "",
            "You asked:",
            "",
            "> " + question,
            "",
            "This application demonstrates a **Full-Stack AI Integration**.",
            "",
            "### Request Flow",
            "",
            "1. User enters a question.",
            "2. Next.js sends a POST request.",
            "3. FastAPI receives the request.",
            "4. The backend generates an answer.",
            "5. The answer is streamed progressively.",
            "6. React displays the response.",
            "7. Markdown is rendered in the browser.",
            "8. Sources are displayed below the answer.",
            "",
            "### Architecture",
            "",
            "```text",
            "User",
            "  |",
            "  v",
            "Next.js",
            "  |",
            "  | POST /ask",
            "  v",
            "FastAPI",
            "  |",
            "  v",
            "AI Response Generator",
            "  |",
            "  v",
            "StreamingResponse",
            "  |",
            "  v",
            "ReadableStream",
            "  |",
            "  v",
            "React UI",
            "```",
            "",
            "This project does **not require an OpenAI API key**."
        ])


        sources = [
            {
                "title": "FastAPI Documentation",
                "url": "https://fastapi.tiangolo.com/"
            },
            {
                "title": "Next.js Documentation",
                "url": "https://nextjs.org/docs"
            }
        ]


    return answer, sources


# ============================================================
# STREAMING RESPONSE
# ============================================================

async def stream_response(
    question: str,
    history: list
):

    answer, sources = generate_answer(question)


    # Send answer progressively
    words = answer.split(" ")


    for index, word in enumerate(words):

        chunk = word

        if index < len(words) - 1:
            chunk += " "


        data = {
            "type": "token",
            "content": chunk
        }


        yield json.dumps(data) + "\n"


        # Small delay so streaming can be seen
        await asyncio.sleep(0.04)


    # Send sources after answer
    source_data = {
        "type": "sources",
        "sources": sources
    }


    yield json.dumps(source_data) + "\n"


# ============================================================
# POST /ask
# ============================================================

@app.post("/ask")
async def ask(request: AskRequest):

    question = request.question.strip()


    # Empty question
    if not question:

        raise HTTPException(
            status_code=400,
            detail="Please enter a question before sending your message."
        )


    # Question too long
    if len(question) > 2000:

        raise HTTPException(
            status_code=400,
            detail="Your question is too long. Please keep it under 2000 characters."
        )


    return StreamingResponse(
        stream_response(
            question,
            request.history
        ),
        media_type="application/x-ndjson"
    )


# ============================================================
# HEALTH CHECK
# ============================================================

@app.get("/")
def root():

    return {
        "status": "online",
        "message": "Full Stack AI Assistant API is running"
    }


# ============================================================
# SERVER HEALTH
# ============================================================

@app.get("/health")
def health():

    return {
        "status": "healthy"
    }
````

````python
%%writefile /content/fullstack-ai/frontend/src/app/page.tsx

"use client";

import { FormEvent, useState } from "react";
import ReactMarkdown from "react-markdown";

interface Source {
    title: string;
    url: string;
}

interface Message {
    role: "user" | "assistant";
    content: string;
    sources?: Source[];
}

export default function Home() {

    const [messages, setMessages] = useState<Message[]>([]);
    const [input, setInput] = useState("");
    const [loading, setLoading] = useState(false);
    const [error, setError] = useState("");

    async function sendMessage(event: FormEvent) {

        event.preventDefault();

        const question = input.trim();

        if (!question || loading) {
            return;
        }

        setError("");
        setInput("");
        setLoading(true);

        const userMessage: Message = {
            role: "user",
            content: question
        };

        const assistantMessage: Message = {
            role: "assistant",
            content: "",
            sources: []
        };

        const updatedMessages = [
            ...messages,
            userMessage
        ];

        setMessages([
            ...updatedMessages,
            assistantMessage
        ]);

        try {

            const response = await fetch(
                "http://127.0.0.1:8000/ask",
                {
                    method: "POST",

                    headers: {
                        "Content-Type": "application/json"
                    },

                    body: JSON.stringify({
                        question: question,
                        history: messages
                    })
                }
            );

            if (!response.ok) {

                let errorMessage =
                    "The AI service could not process your request.";

                try {

                    const errorData = await response.json();

                    if (errorData.detail) {
                        errorMessage = errorData.detail;
                    }

                } catch {
                    // Use default message
                }

                throw new Error(errorMessage);
            }

            if (!response.body) {

                throw new Error(
                    "The server did not return a streaming response."
                );
            }

            const reader = response.body.getReader();

            const decoder = new TextDecoder();

            let buffer = "";

            while (true) {

                const { value, done } =
                    await reader.read();

                if (done) {
                    break;
                }

                buffer += decoder.decode(value, {
                    stream: true
                });

                const lines = buffer.split("\n");

                buffer = lines.pop() || "";

                for (const line of lines) {

                    if (!line.trim()) {
                        continue;
                    }

                    try {

                        const data = JSON.parse(line);

                        // -------------------------
                        // STREAMING TOKEN
                        // -------------------------

                        if (data.type === "token") {

                            setMessages(previous => {

                                const copy = [...previous];

                                const lastIndex =
                                    copy.length - 1;

                                copy[lastIndex] = {
                                    ...copy[lastIndex],
                                    content:
                                        copy[lastIndex].content +
                                        data.content
                                };

                                return copy;
                            });
                        }

                        // -------------------------
                        // SOURCES
                        // -------------------------

                        if (data.type === "sources") {

                            setMessages(previous => {

                                const copy = [...previous];

                                const lastIndex =
                                    copy.length - 1;

                                copy[lastIndex] = {
                                    ...copy[lastIndex],
                                    sources: data.sources
                                };

                                return copy;
                            });
                        }

                    } catch (parseError) {

                        console.error(
                            "Stream parsing error:",
                            parseError
                        );
                    }
                }
            }

        } catch (err) {

            const message =
                err instanceof Error
                    ? err.message
                    : "Unable to connect to the AI server.";

            setError(message);

            setMessages(previous => {

                const last = previous[previous.length - 1];

                if (
                    last &&
                    last.role === "assistant" &&
                    last.content === ""
                ) {
                    return previous.slice(0, -1);
                }

                return previous;
            });

        } finally {

            setLoading(false);
        }
    }


    return (
        <main className="app">

            <section className="chat-container">

                {/* HEADER */}

                <header className="header">

                    <h1>AI Assistant</h1>

                    <p>
                        Full-Stack AI Assistant
                    </p>

                </header>


                {/* CHAT HISTORY */}

                <div className="messages">

                    {messages.length === 0 && (

                        <div className="welcome">

                            <h2>
                                How can I help you?
                            </h2>

                            <p>
                                Ask a question to start a conversation.
                            </p>

                        </div>
                    )}


                    {messages.map(
                        (message, index) => (

                            <div
                                key={index}
                                className={
                                    `message-row ${message.role}`
                                }
                            >

                                <div className="message">

                                    <div className="role">

                                        {message.role === "user"
                                            ? "You"
                                            : "AI"}

                                    </div>


                                    {message.role === "assistant" ? (

                                        <div className="markdown">

                                            <ReactMarkdown>
                                                {message.content}
                                            </ReactMarkdown>

                                        </div>

                                    ) : (

                                        <div className="user-text">
                                            {message.content}
                                        </div>

                                    )}


                                    {/* SOURCES */}

                                    {message.role === "assistant" &&
                                        message.sources &&
                                        message.sources.length > 0 && (

                                            <details className="sources">

                                                <summary>
                                                    Sources (
                                                    {message.sources.length}
                                                    )
                                                </summary>

                                                <div className="source-list">

                                                    {message.sources.map(
                                                        (source, sourceIndex) => (

                                                            <a
                                                                key={sourceIndex}
                                                                href={source.url}
                                                                target="_blank"
                                                                rel="noopener noreferrer"
                                                                className="source"
                                                            >

                                                                <strong>
                                                                    {source.title}
                                                                </strong>

                                                                <span>
                                                                    {source.url}
                                                                </span>

                                                            </a>

                                                        )
                                                    )}

                                                </div>

                                            </details>

                                        )}

                                </div>

                            </div>

                        )
                    )}


                    {/* TYPING INDICATOR */}

                    {loading &&
                        messages.length > 0 &&
                        messages[messages.length - 1].content === "" && (

                            <div className="typing">

                                <span></span>
                                <span></span>
                                <span></span>

                                <label>
                                    AI is thinking...
                                </label>

                            </div>
                        )}

                </div>


                {/* ERROR */}

                {error && (

                    <div className="error">

                        <strong>
                            Request failed:
                        </strong>

                        <span>
                            {error}
                        </span>

                    </div>
                )}


                {/* INPUT */}

                <form
                    className="input-area"
                    onSubmit={sendMessage}
                >

                    <textarea
                        value={input}
                        onChange={(event) =>
                            setInput(event.target.value)
                        }
                        placeholder="Ask anything..."
                        disabled={loading}
                        rows={1}
                    />

                    <button
                        type="submit"
                        disabled={
                            loading ||
                            !input.trim()
                        }
                    >
                        {loading ? "..." : "Send"}
                    </button>

                </form>

            </section>

        </main>
    );
}
````

````python
%%writefile /content/fullstack-ai/frontend/src/app/page.tsx

"use client";

import { FormEvent, useState } from "react";
import ReactMarkdown from "react-markdown";

interface Source {
    title: string;
    url: string;
}

interface Message {
    role: "user" | "assistant";
    content: string;
    sources?: Source[];
}

const BACKEND_URL =
    "https://calculator-coupon-excellent-icq.trycloudflare.com";


export default function Home() {

    const [messages, setMessages] = useState<Message[]>([]);
    const [input, setInput] = useState("");
    const [loading, setLoading] = useState(false);
    const [error, setError] = useState("");


    async function sendMessage(event: FormEvent) {

        event.preventDefault();

        const question = input.trim();

        if (!question || loading) {
            return;
        }

        setError("");
        setInput("");
        setLoading(true);


        const userMessage: Message = {
            role: "user",
            content: question
        };


        const assistantMessage: Message = {
            role: "assistant",
            content: "",
            sources: []
        };


        setMessages(previous => [
            ...previous,
            userMessage,
            assistantMessage
        ]);


        try {

            const response = await fetch(
                BACKEND_URL + "/ask",
                {
                    method: "POST",

                    headers: {
                        "Content-Type": "application/json"
                    },

                    body: JSON.stringify({
                        question: question,
                        history: messages
                    })
                }
            );


            if (!response.ok) {

                let errorMessage =
                    "The AI service could not process your request.";

                try {

                    const errorData =
                        await response.json();

                    if (errorData.detail) {
                        errorMessage =
                            errorData.detail;
                    }

                } catch {

                    // Use default error
                }

                throw new Error(errorMessage);
            }


            if (!response.body) {

                throw new Error(
                    "The server did not return a streaming response."
                );
            }


            const reader =
                response.body.getReader();

            const decoder =
                new TextDecoder();

            let buffer = "";


            while (true) {

                const { value, done } =
                    await reader.read();


                if (done) {
                    break;
                }


                buffer += decoder.decode(
                    value,
                    { stream: true }
                );


                const lines =
                    buffer.split("\n");


                buffer =
                    lines.pop() || "";


                for (const line of lines) {

                    if (!line.trim()) {
                        continue;
                    }


                    try {

                        const data =
                            JSON.parse(line);


                        // -------------------------
                        // TOKEN
                        // -------------------------

                        if (data.type === "token") {

                            setMessages(previous => {

                                const copy =
                                    [...previous];

                                const lastIndex =
                                    copy.length - 1;


                                copy[lastIndex] = {
                                    ...copy[lastIndex],

                                    content:
                                        copy[lastIndex]
                                            .content +
                                        data.content
                                };


                                return copy;
                            });
                        }


                        // -------------------------
                        // SOURCES
                        // -------------------------

                        if (data.type === "sources") {

                            setMessages(previous => {

                                const copy =
                                    [...previous];

                                const lastIndex =
                                    copy.length - 1;


                                copy[lastIndex] = {
                                    ...copy[lastIndex],

                                    sources:
                                        data.sources
                                };


                                return copy;
                            });
                        }

                    } catch (error) {

                        console.error(
                            "Stream parsing error:",
                            error
                        );
                    }
                }
            }

        } catch (error) {

            const message =
                error instanceof Error
                    ? error.message
                    : "Unable to connect to the AI server.";


            setError(message);


            setMessages(previous => {

                const last =
                    previous[previous.length - 1];


                if (
                    last &&
                    last.role === "assistant" &&
                    last.content === ""
                ) {

                    return previous.slice(0, -1);
                }


                return previous;
            });

        } finally {

            setLoading(false);
        }
    }


    return (

        <main className="app">

            <section className="chat-container">


                {/* HEADER */}

                <header className="header">

                    <h1>
                        AI Assistant
                    </h1>

                    <p>
                        Full-Stack AI Assistant
                    </p>

                </header>


                {/* CHAT */}

                <div className="messages">


                    {messages.length === 0 && (

                        <div className="welcome">

                            <h2>
                                How can I help you?
                            </h2>

                            <p>
                                Ask a question to start
                                a conversation.
                            </p>

                        </div>

                    )}


                    {messages.map(
                        (message, index) => (

                            <div
                                key={index}
                                className={
                                    "message-row " +
                                    message.role
                                }
                            >

                                <div className="message">


                                    <div className="role">

                                        {message.role === "user"
                                            ? "You"
                                            : "AI"}

                                    </div>


                                    {message.role === "assistant" ? (

                                        <div className="markdown">

                                            <ReactMarkdown>
                                                {message.content}
                                            </ReactMarkdown>

                                        </div>

                                    ) : (

                                        <div className="user-text">
                                            {message.content}
                                        </div>

                                    )}


                                    {/* SOURCES */}

                                    {message.role === "assistant" &&
                                        message.sources &&
                                        message.sources.length > 0 && (

                                            <details className="sources">

                                                <summary>
                                                    Sources (
                                                    {message.sources.length}
                                                    )
                                                </summary>


                                                <div className="source-list">

                                                    {message.sources.map(
                                                        (
                                                            source,
                                                            sourceIndex
                                                        ) => (

                                                            <a
                                                                key={
                                                                    sourceIndex
                                                                }
                                                                href={
                                                                    source.url
                                                                }
                                                                target="_blank"
                                                                rel="noopener noreferrer"
                                                                className="source"
                                                            >

                                                                <strong>
                                                                    {
                                                                        source.title
                                                                    }
                                                                </strong>

                                                                <span>
                                                                    {
                                                                        source.url
                                                                    }
                                                                </span>

                                                            </a>

                                                        )
                                                    )}

                                                </div>

                                            </details>

                                        )}

                                </div>

                            </div>

                        )
                    )}


                    {/* LOADING */}

                    {loading &&
                        messages.length > 0 &&
                        messages[
                            messages.length - 1
                        ].content === "" && (

                            <div className="typing">

                                <span></span>
                                <span></span>
                                <span></span>

                                <label>
                                    AI is thinking...
                                </label>

                            </div>

                        )}

                </div>


                {/* ERROR */}

                {error && (

                    <div className="error">

                        <strong>
                            Request failed:
                        </strong>

                        <span>
                            {error}
                        </span>

                    </div>

                )}


                {/* INPUT */}

                <form
                    className="input-area"
                    onSubmit={sendMessage}
                >

                    <textarea
                        value={input}
                        onChange={
                            event =>
                                setInput(
                                    event.target.value
                                )
                        }
                        placeholder="Ask anything..."
                        disabled={loading}
                        rows={1}
                    />


                    <button
                        type="submit"
                        disabled={
                            loading ||
                            !input.trim()
                        }
                    >

                        {loading
                            ? "..."
                            : "Send"}

                    </button>

                </form>


            </section>

        </main>
    );
}
````

````python
%%writefile /content/fullstack-ai/frontend/src/app/page.tsx

"use client";

import { useState, FormEvent } from "react";
import ReactMarkdown from "react-markdown";

const BACKEND_URL = "https://calculator-coupon-excellent-icq.trycloudflare.com";

type Source = {
    title: string;
    url: string;
};

type Message = {
    role: "user" | "assistant";
    content: string;
    sources?: Source[];
};

export default function Home() {

    const [messages, setMessages] = useState<Message[]>([]);
    const [input, setInput] = useState("");
    const [loading, setLoading] = useState(false);
    const [error, setError] = useState("");


    async function sendMessage(e: FormEvent<HTMLFormElement>) {

        e.preventDefault();

        const question = input.trim();

        if (!question || loading) {
            return;
        }

        console.log("1. Sending:", question);

        setInput("");
        setError("");
        setLoading(true);


        // Add user message
        setMessages(prev => [
            ...prev,
            {
                role: "user",
                content: question
            },
            {
                role: "assistant",
                content: ""
            }
        ]);


        try {

            console.log(
                "2. Calling:",
                BACKEND_URL + "/ask"
            );


            const response = await fetch(
                BACKEND_URL + "/ask",
                {
                    method: "POST",

                    headers: {
                        "Content-Type": "application/json"
                    },

                    body: JSON.stringify({
                        question: question,
                        history: messages
                    })
                }
            );


            console.log(
                "3. Status:",
                response.status
            );


            if (!response.ok) {

                let message =
                    "The server could not process your question.";

                try {

                    const data =
                        await response.json();

                    if (data.detail) {
                        message = data.detail;
                    }

                } catch {
                    // Keep default error
                }

                throw new Error(message);
            }


            if (!response.body) {

                throw new Error(
                    "The server did not provide a streaming response."
                );
            }


            console.log("4. Stream connected");


            const reader =
                response.body.getReader();

            const decoder =
                new TextDecoder();

            let buffer = "";


            while (true) {

                const result =
                    await reader.read();


                if (result.done) {

                    console.log(
                        "5. Stream finished"
                    );

                    break;
                }


                const text =
                    decoder.decode(
                        result.value,
                        {
                            stream: true
                        }
                    );


                console.log(
                    "Received:",
                    text
                );


                buffer += text;


                const lines =
                    buffer.split("\n");


                buffer =
                    lines.pop() || "";


                for (const line of lines) {

                    if (!line.trim()) {
                        continue;
                    }


                    const data =
                        JSON.parse(line);


                    // -------------------------
                    // TOKEN
                    // -------------------------

                    if (data.type === "token") {

                        setMessages(prev => {

                            const copy =
                                [...prev];

                            const last =
                                copy.length - 1;


                            copy[last] = {
                                ...copy[last],

                                content:
                                    copy[last].content +
                                    data.content
                            };


                            return copy;
                        });
                    }


                    // -------------------------
                    // SOURCES
                    // -------------------------

                    if (data.type === "sources") {

                        setMessages(prev => {

                            const copy =
                                [...prev];

                            const last =
                                copy.length - 1;


                            copy[last] = {
                                ...copy[last],

                                sources:
                                    data.sources
                            };


                            return copy;
                        });
                    }
                }
            }


        } catch (err) {

            console.error(
                "ERROR:",
                err
            );


            const message =
                err instanceof Error
                    ? err.message
                    : "Unable to connect to the server.";


            setError(message);


            // Remove empty assistant message
            setMessages(prev => {

                const copy = [...prev];

                const last =
                    copy[copy.length - 1];


                if (
                    last &&
                    last.role === "assistant" &&
                    last.content === ""
                ) {
                    copy.pop();
                }


                return copy;
            });


        } finally {

            setLoading(false);

        }
    }


    return (

        <main className="app">

            <section className="chat-container">


                {/* HEADER */}

                <header className="header">

                    <h1>
                        AI Assistant
                    </h1>

                    <p>
                        Full-Stack AI Assistant
                    </p>

                </header>


                {/* MESSAGES */}

                <div className="messages">


                    {messages.length === 0 && (

                        <div className="welcome">

                            <h2>
                                How can I help you?
                            </h2>

                            <p>
                                Ask a question to start a conversation.
                            </p>

                        </div>

                    )}


                    {messages.map(
                        (message, index) => (

                            <div
                                key={index}
                                className={
                                    "message-row " +
                                    message.role
                                }
                            >

                                <div className="message">


                                    <div className="role">

                                        {message.role === "user"
                                            ? "You"
                                            : "AI"}

                                    </div>


                                    {/* ASSISTANT */}

                                    {message.role === "assistant" ? (

                                        message.content ? (

                                            <div className="markdown">

                                                <ReactMarkdown>
                                                    {message.content}
                                                </ReactMarkdown>

                                            </div>

                                        ) : (

                                            loading && (

                                                <div className="thinking">

                                                    <span>
                                                        ●
                                                    </span>

                                                    <span>
                                                        ●
                                                    </span>

                                                    <span>
                                                        ●
                                                    </span>

                                                    <b>
                                                        AI is thinking...
                                                    </b>

                                                </div>

                                            )

                                        )

                                    ) : (

                                        <div className="user-text">
                                            {message.content}
                                        </div>

                                    )}


                                    {/* SOURCES */}

                                    {message.role === "assistant" &&
                                        message.sources &&
                                        message.sources.length > 0 && (

                                            <details className="sources">

                                                <summary>
                                                    Sources (
                                                    {message.sources.length}
                                                    )
                                                </summary>


                                                {message.sources.map(
                                                    (source, i) => (

                                                        <a
                                                            key={i}
                                                            href={source.url}
                                                            target="_blank"
                                                            rel="noopener noreferrer"
                                                            className="source"
                                                        >

                                                            <strong>
                                                                {source.title}
                                                            </strong>

                                                            <small>
                                                                {source.url}
                                                            </small>

                                                        </a>

                                                    )
                                                )}

                                            </details>

                                        )}

                                </div>

                            </div>

                        )
                    )}

                </div>


                {/* ERROR */}

                {error && (

                    <div className="error">

                        <strong>
                            Error:
                        </strong>

                        <span>
                            {error}
                        </span>

                    </div>

                )}


                {/* INPUT */}

                <form
                    className="input-area"
                    onSubmit={sendMessage}
                >

                    <textarea
                        value={input}
                        onChange={
                            e =>
                                setInput(e.target.value)
                        }
                        placeholder="Ask anything..."
                        disabled={loading}
                        rows={1}
                    />


                    <button
                        type="submit"
                        disabled={loading}
                    >

                        {loading
                            ? "Sending..."
                            : "Send"}

                    </button>

                </form>


            </section>

        </main>
    );
}
````

````python
%%writefile /content/fullstack-ai/frontend/src/app/api/ask/route.ts

import { NextRequest } from "next/server";

const BACKEND_URL =
    "https://calculator-coupon-excellent-icq.trycloudflare.com";


export async function POST(request: NextRequest) {

    try {

        const body = await request.json();


        const response = await fetch(
            BACKEND_URL + "/ask",
            {
                method: "POST",

                headers: {
                    "Content-Type": "application/json"
                },

                body: JSON.stringify(body),

                cache: "no-store"
            }
        );


        if (!response.ok) {

            const errorText =
                await response.text();

            return new Response(
                errorText || "Backend request failed",
                {
                    status: response.status,
                    headers: {
                        "Content-Type":
                            "application/json"
                    }
                }
            );
        }


        if (!response.body) {

            return new Response(
                JSON.stringify({
                    detail:
                        "Backend did not return a streaming response."
                }),
                {
                    status: 500,
                    headers: {
                        "Content-Type":
                            "application/json"
                    }
                }
            );
        }


        // Pass the backend stream directly
        // to the browser.

        return new Response(
            response.body,
            {
                status: 200,

                headers: {
                    "Content-Type":
                        "application/x-ndjson",

                    "Cache-Control":
                        "no-cache, no-transform",

                    "X-Accel-Buffering":
                        "no"
                }
            }
        );


    } catch (error) {

        console.error(
            "Proxy error:",
            error
        );


        return new Response(
            JSON.stringify({
                detail:
                    "Unable to connect to the AI backend. Please make sure the FastAPI server is running."
            }),
            {
                status: 502,

                headers: {
                    "Content-Type":
                        "application/json"
                }
            }
        );
    }
}
````

````python
import os
from pathlib import Path

FRONTEND = Path("/content/fullstack-ai/frontend")
APP = FRONTEND / "src/app"
API = APP / "api/ask"

API.mkdir(parents=True, exist_ok=True)

# ============================================================
# 1. BACKEND URL FOR NEXT.JS SERVER
# ============================================================

backend_url = "https://calculator-coupon-excellent-icq.trycloudflare.com"

(FRONTEND / ".env.local").write_text(
    f"BACKEND_URL={backend_url}\n"
)


# ============================================================
# 2. NEXT.JS API PROXY
# ============================================================

route_code = r'''
import { NextRequest } from "next/server";

const BACKEND_URL = process.env.BACKEND_URL;

export async function POST(request: NextRequest) {

    if (!BACKEND_URL) {
        return new Response(
            JSON.stringify({
                detail: "Backend URL is not configured."
            }),
            {
                status: 500,
                headers: {
                    "Content-Type": "application/json"
                }
            }
        );
    }

    try {

        const body = await request.json();

        const response = await fetch(
            `${BACKEND_URL}/ask`,
            {
                method: "POST",

                headers: {
                    "Content-Type": "application/json"
                },

                body: JSON.stringify(body),

                cache: "no-store"
            }
        );

        if (!response.ok) {

            const text = await response.text();

            return new Response(
                text || JSON.stringify({
                    detail: "FastAPI returned an error."
                }),
                {
                    status: response.status,
                    headers: {
                        "Content-Type": "application/json"
                    }
                }
            );
        }

        if (!response.body) {

            return new Response(
                JSON.stringify({
                    detail:
                        "FastAPI did not return a streaming response."
                }),
                {
                    status: 500,
                    headers: {
                        "Content-Type": "application/json"
                    }
                }
            );
        }

        return new Response(
            response.body,
            {
                status: 200,
                headers: {
                    "Content-Type": "application/x-ndjson",
                    "Cache-Control": "no-cache, no-transform"
                }
            }
        );

    } catch (error) {

        console.error("Proxy error:", error);

        return new Response(
            JSON.stringify({
                detail:
                    "Unable to connect to the FastAPI backend."
            }),
            {
                status: 502,
                headers: {
                    "Content-Type": "application/json"
                }
            }
        );
    }
}
'''

(API / "route.ts").write_text(route_code)


# ============================================================
# 3. CORRECT REACT FRONTEND
# ============================================================

page_code = r'''
"use client";

import { FormEvent, useState } from "react";
import ReactMarkdown from "react-markdown";

type Source = {
    title: string;
    url: string;
};

type Message = {
    role: "user" | "assistant";
    content: string;
    sources?: Source[];
};

export default function Home() {

    const [messages, setMessages] =
        useState<Message[]>([]);

    const [input, setInput] =
        useState("");

    const [loading, setLoading] =
        useState(false);

    const [error, setError] =
        useState("");


    async function sendMessage(
        event: FormEvent<HTMLFormElement>
    ) {

        event.preventDefault();

        const question = input.trim();

        if (!question || loading) {
            return;
        }

        console.log("Sending:", question);

        setInput("");
        setError("");
        setLoading(true);


        const userMessage: Message = {
            role: "user",
            content: question
        };

        const assistantMessage: Message = {
            role: "assistant",
            content: "",
            sources: []
        };


        setMessages(previous => [
            ...previous,
            userMessage,
            assistantMessage
        ]);


        try {

            /*
             * IMPORTANT:
             *
             * Browser calls Next.js only.
             *
             * Next.js /api/ask calls FastAPI.
             *
             * Therefore there is NO browser CORS problem.
             */

            console.log("Calling: /api/ask");


            const response = await fetch(
                "/api/ask",
                {
                    method: "POST",

                    headers: {
                        "Content-Type": "application/json"
                    },

                    body: JSON.stringify({
                        question: question,
                        history: messages
                    })
                }
            );


            console.log(
                "Response status:",
                response.status
            );


            if (!response.ok) {

                let message =
                    "The AI service could not process your request.";

                try {

                    const data =
                        await response.json();

                    if (data.detail) {
                        message = data.detail;
                    }

                } catch {
                    // Keep readable default
                }

                throw new Error(message);
            }


            if (!response.body) {

                throw new Error(
                    "The server did not provide a streaming response."
                );
            }


            console.log(
                "Streaming connection established."
            );


            const reader =
                response.body.getReader();

            const decoder =
                new TextDecoder();

            let buffer = "";


            while (true) {

                const { value, done } =
                    await reader.read();


                if (done) {
                    break;
                }


                buffer += decoder.decode(
                    value,
                    {
                        stream: true
                    }
                );


                const lines =
                    buffer.split("\n");


                buffer =
                    lines.pop() || "";


                for (const line of lines) {

                    if (!line.trim()) {
                        continue;
                    }


                    try {

                        const data =
                            JSON.parse(line);


                        /*
                         * TOKEN
                         */

                        if (data.type === "token") {

                            setMessages(previous => {

                                const copy =
                                    [...previous];

                                const lastIndex =
                                    copy.length - 1;


                                copy[lastIndex] = {
                                    ...copy[lastIndex],

                                    content:
                                        copy[lastIndex].content +
                                        data.content
                                };


                                return copy;
                            });
                        }


                        /*
                         * SOURCES
                         */

                        if (data.type === "sources") {

                            setMessages(previous => {

                                const copy =
                                    [...previous];

                                const lastIndex =
                                    copy.length - 1;


                                copy[lastIndex] = {
                                    ...copy[lastIndex],

                                    sources:
                                        data.sources
                                };


                                return copy;
                            });
                        }

                    } catch (parseError) {

                        console.error(
                            "Stream parsing error:",
                            parseError
                        );
                    }
                }
            }


            console.log(
                "Streaming completed."
            );


        } catch (error) {

            console.error(
                "Request error:",
                error
            );


            const message =
                error instanceof Error
                    ? error.message
                    : "Unable to connect to the AI service.";


            setError(message);


            setMessages(previous => {

                const copy =
                    [...previous];

                const last =
                    copy[copy.length - 1];


                if (
                    last &&
                    last.role === "assistant" &&
                    last.content === ""
                ) {

                    copy.pop();
                }


                return copy;
            });


        } finally {

            setLoading(false);
        }
    }


    return (

        <main className="app">

            <section className="chat-container">


                {/* HEADER */}

                <header className="header">

                    <h1>
                        AI Assistant
                    </h1>

                    <p>
                        Full-Stack AI Assistant
                    </p>

                </header>


                {/* MESSAGE HISTORY */}

                <div className="messages">


                    {messages.length === 0 && (

                        <div className="welcome">

                            <h2>
                                How can I help you?
                            </h2>

                            <p>
                                Ask a question to start
                                a conversation.
                            </p>

                        </div>

                    )}


                    {messages.map(
                        (message, index) => (

                            <div
                                key={index}
                                className={
                                    "message-row " +
                                    message.role
                                }
                            >

                                <div className="message">


                                    <div className="role">

                                        {message.role === "user"
                                            ? "You"
                                            : "AI"}

                                    </div>


                                    {message.role === "assistant" ? (

                                        message.content ? (

                                            <div className="markdown">

                                                <ReactMarkdown>
                                                    {message.content}
                                                </ReactMarkdown>

                                            </div>

                                        ) : (

                                            loading && (

                                                <div className="typing">

                                                    <span>●</span>
                                                    <span>●</span>
                                                    <span>●</span>

                                                    <label>
                                                        AI is thinking...
                                                    </label>

                                                </div>

                                            )

                                        )

                                    ) : (

                                        <div className="user-text">
                                            {message.content}
                                        </div>

                                    )}


                                    {/* SOURCES */}

                                    {message.role === "assistant" &&
                                        message.sources &&
                                        message.sources.length > 0 && (

                                            <details className="sources">

                                                <summary>
                                                    Sources (
                                                    {message.sources.length}
                                                    )
                                                </summary>


                                                <div className="source-list">

                                                    {message.sources.map(
                                                        (source, i) => (

                                                            <a
                                                                key={i}
                                                                href={source.url}
                                                                target="_blank"
                                                                rel="noopener noreferrer"
                                                                className="source"
                                                            >

                                                                <strong>
                                                                    {source.title}
                                                                </strong>

                                                                <span>
                                                                    {source.url}
                                                                </span>

                                                            </a>

                                                        )
                                                    )}

                                                </div>

                                            </details>

                                        )}

                                </div>

                            </div>

                        )
                    )}

                </div>


                {/* ERROR */}

                {error && (

                    <div className="error">

                        <strong>
                            Error:
                        </strong>

                        <span>
                            {error}
                        </span>

                    </div>

                )}


                {/* INPUT */}

                <form
                    className="input-area"
                    onSubmit={sendMessage}
                >

                    <textarea
                        value={input}
                        onChange={
                            event =>
                                setInput(event.target.value)
                        }
                        placeholder="Ask anything..."
                        disabled={loading}
                        rows={1}
                    />


                    <button
                        type="submit"
                        disabled={loading}
                    >

                        {loading
                            ? "Sending..."
                            : "Send"}

                    </button>

                </form>

            </section>

        </main>
    );
}
'''

(APP / "page.tsx").write_text(page_code)

print("============================================")
print("✅ FRONTEND RESET COMPLETE")
print("============================================")
print()
print("Created:")
print("  ✓ .env.local")
print("  ✓ src/app/api/ask/route.ts")
print("  ✓ src/app/page.tsx")
print()
print("Browser will now call:")
print("  /api/ask")
print()
print("Next.js server will call:")
print(f"  {backend_url}/ask")
````

````python
import re

with open("/content/frontend-tunnel.log", "r") as f:
    log = f.read()

urls = re.findall(
    r"https://[a-zA-Z0-9.-]+\.trycloudflare\.com",
    log
)

if urls:
    FRONTEND_URL = urls[-1]
    print("✅ FINAL FRONTEND URL:")
    print(FRONTEND_URL)
else:
    print("❌ No tunnel URL found")
    print(log)
````

````python
import requests
import json

r = requests.post(
    "http://127.0.0.1:3000/api/ask",
    json={
        "question": "What is FastAPI?",
        "history": []
    },
    stream=True,
    timeout=30
)

for line in r.iter_lines():
    if line:
        data = json.loads(line.decode())

        if data["type"] == "sources":
            print("✅ SOURCES RECEIVED:")
            print(json.dumps(data["sources"], indent=2))
````
