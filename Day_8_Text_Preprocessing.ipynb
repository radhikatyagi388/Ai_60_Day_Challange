```python
import nltk
import re
import string

from nltk.tokenize import word_tokenize, sent_tokenize
from nltk.corpus import stopwords
from nltk.stem import PorterStemmer, WordNetLemmatizer

# Download required NLTK resources
nltk.download('punkt')
nltk.download('punkt_tab')
nltk.download('stopwords')
nltk.download('wordnet')
nltk.download('omw-1.4')

print("Libraries loaded successfully!")
```

```python
samples = {
    "Social Media": """
OMG!!! AI is AMAZING 🤖🔥!!! I am learning NLP today, and it's REALLY exciting!!!
Can't wait to build more AI projects 😍 #AI #MachineLearning
""",

    "News Article": """
Artificial Intelligence is transforming industries around the world.
Companies are using AI to improve customer service, automate repetitive tasks,
and analyze large amounts of data. However, experts say that responsible AI
development is extremely important.
""",

    "Customer Review": """
I LOVED this product!!! 😍 The design is beautiful, the performance is AMAZING,
and the delivery was super fast. However, the packaging was not great...
I would definitely recommend this product to my friends!!!
"""
}

for name, text in samples.items():
    print("=" * 70)
    print(name)
    print("=" * 70)
    print(text)
```

```python
for name, text in samples.items():

    print("\n" + "=" * 70)
    print(name)
    print("=" * 70)

    # Sentence-level tokenization
    sentences = sent_tokenize(text)

    print("\nSentence Tokens:")
    for sentence in sentences:
        print(sentence)

    # Word-level tokenization
    words = word_tokenize(text)

    print("\nWord Tokens:")
    print(words)
```

```python
def clean_text(text):

    # Convert text to lowercase so "AI", "Ai", and "ai" are treated as the same word.
    text = text.lower()

    # Remove emojis and non-ASCII characters.
    text = text.encode("ascii", "ignore").decode()

    # Remove punctuation using regex.
    # This is necessary because punctuation such as !, ?, and , usually
    # does not carry useful information for basic word-level NLP processing.
    text = re.sub(r'[^\w\s]', '', text)

    # Remove extra whitespace
    text = re.sub(r'\s+', ' ', text).strip()

    return text


for name, text in samples.items():

    cleaned = clean_text(text)

    print("\n" + "=" * 70)
    print(name)
    print("=" * 70)

    print("BEFORE:")
    print(text)

    print("\nAFTER:")
    print(cleaned)
```

```python
tokenized_samples = {}

for name, text in samples.items():

    cleaned = clean_text(text)

    tokens = word_tokenize(cleaned)

    tokenized_samples[name] = tokens

    print("\n" + "=" * 70)
    print(name)
    print("=" * 70)

    print("Tokens:")
    print(tokens)
```

```python
stop_words = set(stopwords.words('english'))

for name, tokens in tokenized_samples.items():

    filtered_tokens = [
        word for word in tokens
        if word not in stop_words
    ]

    print("\n" + "=" * 70)
    print(name)
    print("=" * 70)

    print("Before stopword removal:")
    print(tokens)

    print("\nAfter stopword removal:")
    print(filtered_tokens)

    print("\nVocabulary size before:", len(set(tokens)))
    print("Vocabulary size after:", len(set(filtered_tokens)))
```

```python
stemmer = PorterStemmer()
lemmatizer = WordNetLemmatizer()

comparison_words = [
    "playing",
    "plays",
    "played",
    "studies",
    "studying",
    "running",
    "runs",
    "easily",
    "better",
    "cars"
]

print(f"{'Word':<15} {'Stemmed':<15} {'Lemmatized':<15}")
print("-" * 45)

for word in comparison_words:

    stemmed = stemmer.stem(word)
    lemmatized = lemmatizer.lemmatize(word)

    print(f"{word:<15} {stemmed:<15} {lemmatized:<15}")
```

```python
for name, tokens in tokenized_samples.items():

    filtered_tokens = [
        word for word in tokens
        if word not in stop_words
    ]

    stemmed_words = [
        stemmer.stem(word)
        for word in filtered_tokens
    ]

    lemmatized_words = [
        lemmatizer.lemmatize(word)
        for word in filtered_tokens
    ]

    print("\n" + "=" * 70)
    print(name)
    print("=" * 70)

    print("\nOriginal:")
    print(filtered_tokens)

    print("\nStemmed:")
    print(stemmed_words)

    print("\nLemmatized:")
    print(lemmatized_words)
```

```python
for name, text in samples.items():

    print("\n" + "#" * 80)
    print(f"PROCESSING: {name}")
    print("#" * 80)

    # STEP 1: Sentence tokenization
    sentences = sent_tokenize(text)

    print("\nSTEP 1 — Sentence Tokenization")
    print("Before:")
    print(text.strip())
    print("\nAfter:")
    print(sentences)

    # STEP 2: Word tokenization
    words = word_tokenize(text)

    print("\nSTEP 2 — Word Tokenization")
    print("Before:")
    print(text.strip())
    print("\nAfter:")
    print(words)

    # STEP 3: Punctuation removal
    cleaned = clean_text(text)
    cleaned_tokens = word_tokenize(cleaned)

    print("\nSTEP 3 — Punctuation Removal + Normalization")
    print("Before:")
    print(words)
    print("\nAfter:")
    print(cleaned_tokens)

    # STEP 4: Stopword removal
    filtered = [
        word for word in cleaned_tokens
        if word not in stop_words
    ]

    print("\nSTEP 4 — Stopword Removal")
    print("Before:")
    print(cleaned_tokens)
    print("\nAfter:")
    print(filtered)

    # STEP 5: Stemming and Lemmatization
    stemmed = [
        stemmer.stem(word)
        for word in filtered
    ]

    lemmatized = [
        lemmatizer.lemmatize(word)
        for word in filtered
    ]

    print("\nSTEP 5 — Stemming vs Lemmatization")

    print("Before:")
    print(filtered)

    print("\nStemmed:")
    print(stemmed)

    print("\nLemmatized:")
    print(lemmatized)
```

```python
print("=" * 80)
print("VOCABULARY SIZE COMPARISON")
print("=" * 80)

for name, text in samples.items():

    cleaned = clean_text(text)

    tokens = word_tokenize(cleaned)

    filtered = [
        word for word in tokens
        if word not in stop_words
    ]

    before_vocab = len(set(tokens))
    after_vocab = len(set(filtered))

    print(f"\n{name}")
    print("Vocabulary before stopword removal:", before_vocab)
    print("Vocabulary after stopword removal :", after_vocab)
    print("Reduction:", before_vocab - after_vocab)
```

```python
print("""
RAW TEXT
   ↓
Sentence Tokenization
   ↓
Word Tokenization
   ↓
Lowercase + Punctuation Removal
   ↓
Stopword Removal
   ↓
Stemming / Lemmatization
   ↓
CLEAN STRUCTURED TOKENS
""")
```

```python
What I Learned

Through this experiment, I learned that raw human language cannot be directly treated as clean machine-readable data. Text preprocessing converts noisy text into structured tokens that are easier for NLP algorithms to process.

I observed that tokenization breaks text into meaningful units, while lowercasing and punctuation removal reduce unnecessary variations. Stopword removal can reduce vocabulary size by removing extremely common words. I also observed an important difference between stemming and lemmatization: stemming is faster and more aggressive but can produce incomplete words, while lemmatization generally produces more meaningful dictionary forms.

Skipping preprocessing can cause duplicate or inconsistent tokens, increase vocabulary size, introduce noise, and reduce the quality of downstream NLP models. However, preprocessing should always depend on the NLP task because punctuation, stopwords, and word forms can sometimes contain useful information.
```
