```python
def fixed_size_chunking(text, chunk_size, overlap=0):
    chunks = []

    start = 0

    while start < len(text):
        end = start + chunk_size
        chunks.append(text[start:end])

        if end >= len(text):
            break

        start = end - overlap

    return chunks
```

```python
def recursive_chunking(text, chunk_size, overlap):
    splitter = RecursiveCharacterTextSplitter(
        chunk_size=chunk_size,
        chunk_overlap=overlap,
        separators=["\n\n", "\n", ". ", " ", ""]
    )

    return splitter.split_text(text)
```

```python
import re

def tokenize(text):
    return set(
        re.findall(r'\b[a-zA-Z]{3,}\b', text.lower())
    )
```

```python
def relevance_score(query, chunk):
    query_words = tokenize(query)
    chunk_words = tokenize(chunk)

    if not query_words:
        return 0

    common_words = query_words.intersection(chunk_words)

    return len(common_words)
```

```python
def retrieve_chunks(query, chunks, top_k=3):
    scored_chunks = []

    for i, chunk in enumerate(chunks):
        score = relevance_score(query, chunk)

        scored_chunks.append({
            "index": i,
            "score": score,
            "text": chunk
        })

    scored_chunks.sort(
        key=lambda x: x["score"],
        reverse=True
    )

    return scored_chunks[:top_k]
```

```python
query = "How does retrieval augmented generation improve language model responses?"
```

```python
chunks = recursive_chunking(
    document,
    chunk_size=500,
    overlap=50
)

results = retrieve_chunks(
    query,
    chunks,
    top_k=3
)

for result in results:
    print("Chunk:", result["index"])
    print("Score:", result["score"])
    print(result["text"])
    print("-" * 80)
```

```python
chunk_sizes = [100, 300, 500, 1000]
overlaps = [0, 50, 100]

experiment_results = []

for chunk_size in chunk_sizes:

    for requested_overlap in overlaps:

        overlap = requested_overlap

        # LangChain requires overlap < chunk_size
        if overlap >= chunk_size:
            overlap = chunk_size - 1

        chunks = recursive_chunking(
            document,
            chunk_size,
            overlap
        )

        retrieved = retrieve_chunks(
            query,
            chunks,
            top_k=3
        )

        experiment_results.append({
            "chunk_size": chunk_size,
            "requested_overlap": requested_overlap,
            "actual_overlap": overlap,
            "num_chunks": len(chunks),
            "results": retrieved
        })
```

```python
for result in experiment_results:

    print("\n" + "=" * 80)

    print(
        f"Chunk Size: {result['chunk_size']} | "
        f"Requested Overlap: {result['requested_overlap']} | "
        f"Actual Overlap: {result['actual_overlap']}"
    )

    print(
        f"Total Chunks: {result['num_chunks']}"
    )

    for retrieved in result["results"]:

        print(
            f"\nChunk {retrieved['index']} "
            f"| Score: {retrieved['score']}"
        )

        print(retrieved["text"][:500])
```

```python
def completeness_score(text):
    score = 0

    # Sentence ending
    if text.strip().endswith((".", "!", "?")):
        score += 1

    # Starts reasonably
    if text.strip()[0].isupper():
        score += 1

    # Doesn't look truncated
    if not text.strip().endswith(
        ("the", "a", "an", "of", "to", "and", "or", "with")
    ):
        score += 1

    return score
```

```python
def retrieval_quality(query, retrieved_results):

    if not retrieved_results:
        return 0

    relevance = sum(
        r["score"] for r in retrieved_results
    )

    completeness = sum(
        completeness_score(r["text"])
        for r in retrieved_results
    )

    return relevance + completeness
```

```python
for result in experiment_results:

    quality = retrieval_quality(
        query,
        result["results"]
    )

    result["quality_score"] = quality

    print(
        f"Chunk={result['chunk_size']}, "
        f"Overlap={result['requested_overlap']}, "
        f"Quality={quality}"
    )
```

```python
best_result = max(
    experiment_results,
    key=lambda x: x["quality_score"]
)

print("BEST CONFIGURATION")
print("------------------")

print(
    "Chunk Size:",
    best_result["chunk_size"]
)

print(
    "Requested Overlap:",
    best_result["requested_overlap"]
)

print(
    "Actual Overlap:",
    best_result["actual_overlap"]
)

print(
    "Quality Score:",
    best_result["quality_score"]
)
```

```python
def fixed_chunks_with_positions(text, chunk_size, overlap):
    chunks = []

    start = 0

    while start < len(text):

        end = min(start + chunk_size, len(text))

        chunks.append({
            "start": start,
            "end": end,
            "text": text[start:end]
        })

        if end >= len(text):
            break

        start = end - overlap

    return chunks
```

```python
def find_boundary_failures(text, chunk_size, overlap=0):

    chunks = fixed_chunks_with_positions(
        text,
        chunk_size,
        overlap
    )

    failures = []

    for i in range(len(chunks) - 1):

        boundary = chunks[i]["end"]

        before = text[boundary - 1]
        after = text[boundary]

        # If boundary doesn't occur after punctuation,
        # it may have split a sentence.
        if before not in ".!?":

            failures.append({
                "chunk": i,
                "position": boundary,
                "before": text[boundary-100:boundary],
                "after": text[boundary:boundary+100]
            })

    return failures
```

```python
failures = find_boundary_failures(
    document,
    chunk_size=300,
    overlap=0
)

for i, failure in enumerate(failures[:3]):

    print("\n" + "=" * 80)

    print(
        f"Boundary Failure {i+1}"
    )

    print(
        "Character Position:",
        failure["position"]
    )

    print("\nBefore boundary:")
    print(failure["before"])

    print("\nAfter boundary:")
    print(failure["after"])
```
